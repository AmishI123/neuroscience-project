## 2. numpy — the scientist's calculator

Real experiments have thousands of numbers. **numpy** lets us do math on whole lists of numbers at once, fast. We `import` it (load the tool) and nickname it `np`.

```python
import numpy as np

activity = np.array([0.1, 0.4, 0.9, 0.3, 0.2])

print("The array:", activity)
print("Average (mean):", activity.mean())
print("Biggest value (max):", activity.max())
print("Everything doubled:", activity * 2)   # math happens to every number at once!
```

That last line is the magic: `activity * 2` doubled *every* number in one step. No loop needed. This is why scientists love numpy.

## 3. Let's build a tiny (pretend) brain experiment

Before we use real recorded data (next week), we'll **simulate** a simple experiment so we can learn the tools safely. Here's the setup:

- Over time, a mouse sometimes runs and sometimes sits still.
- We record one neuron in its visual cortex.
- **Secret rule we're building in:** this neuron fires a bit more when the mouse is running (plus some random noise, because brains are noisy).

This is *exactly* the kind of effect our whole project is about — so watch for it.

```python
# We use a fixed random 'seed' so everyone gets the same numbers.
rng = np.random.default_rng(seed=42)

n_timepoints = 200

# Running speed at each moment (mostly small, sometimes the mouse sprints):
running_speed = rng.exponential(scale=1.0, size=n_timepoints)

# The neuron's activity = a baseline + a boost from running + random noise
baseline = 0.5
running_boost = 0.8 * running_speed          # <-- the effect we want to discover later
noise = rng.normal(0, 0.5, size=n_timepoints)
neuron_activity = baseline + running_boost + noise

print("We simulated", n_timepoints, "moments in time.")
print("First 5 running speeds:  ", running_speed[:5].round(2))
print("First 5 neuron activities:", neuron_activity[:5].round(2))
```

## 4. Making our first plots

Numbers are hard to read. **Pictures** reveal patterns. We use `matplotlib` for plotting.

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 3))
plt.plot(neuron_activity)
plt.xlabel("Time (moments)")
plt.ylabel("Neuron activity")
plt.title("Our neuron's activity over time")
plt.show()
```

```python
# Let's overlay running speed and neuron activity to compare them.
plt.figure(figsize=(10, 3))
plt.plot(neuron_activity, label="Neuron activity")
plt.plot(running_speed, label="Running speed", alpha=0.7)
plt.xlabel("Time (moments)")
plt.legend()
plt.title("Do they rise and fall together?")
plt.show()
```

## 5. The big-idea plot: does running relate to firing?

A **scatter plot** puts one value on the x-axis and another on the y-axis. If the neuron really fires more when the mouse runs, the dots should trend **upward** to the right.

```python
plt.figure(figsize=(5, 5))
plt.scatter(running_speed, neuron_activity, alpha=0.5)
plt.xlabel("Running speed")
plt.ylabel("Neuron activity")
plt.title("Each dot is one moment in time")
plt.show()

# A single number summarizing the relationship: the 'correlation'.
# +1 = perfectly rise together, 0 = no relationship, -1 = one up while other down.
correlation = np.corrcoef(running_speed, neuron_activity)[0, 1]
print("Correlation between running and neuron activity:", round(correlation, 2))
```

**Discuss:** The dots trend upward, and the correlation is clearly positive. We *built in* that rule, so finding it confirms our tools work.
